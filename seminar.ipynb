### Списки и циклы

Перед нами средние месячные температуры в Москве за 2025 год, рассчитанные
по ежедневным данным из файла `moscow-weather-2025.csv`.

```
# Значения расположены по порядку: от января до декабря, в °C.
temps = [0.1419, -4.7929, 4.3226, 8.5, 13.2194, 16.9567,
         21.6742, 17.0194, 13.6967, 7.4194, 3.7933, -1.1903]
len(temps)  # длина списка
```

К элементу списка можно обратиться по индексу.
В Python нумерация начинается с нуля: январь — `temps[0]`, февраль —
`temps[1]`, декабрь — `temps[11]` или `temps[-1]`.

```
print("Средняя температура в январе, °C:", temps[0])
```

`print()`, len() — пример функции: она получает значения в скобках (аргументы)
и выводит их. В Python много встроенных функций, например `round()`, возвращает округлённое число

```
print("Средняя температура в январе, °C:", round(temps[0]))
```

Для объединения текста и значений удобны f-строки: поставьте `f` перед
кавычками и поместите выражение в фигурные скобки.

```
print(f"Средняя температура в январе: {temps[0]} °C")
```

Внутри фигурных скобок можно указать формат `:.1f`: он отображает число в определенном формате — с одним знаком после запятой (1 float)

```
print(f"Средняя температура в январе: {temps[0]:.1f} °C")

# print("Средняя температура в январе, °C:", round(temps[0], 1))
```

Цикл `for` позволяет выполнить один и тот же код для каждого элемента списка.
Отступ показывает, какие строки относятся к циклу.

```
for temp in temps:
    print(f"{temp:.1f} °C")
```

Добавим список названий месяцев в том же порядке.

```
months = ["Январь", "Февраль", "Март", "Апрель", "Май", "Июнь",
          "Июль", "Август", "Сентябрь", "Октябрь", "Ноябрь", "Декабрь"]
```

Можно перебрать индексы и взять элементы обоих списков по одному индексу.
`range(12)` даёт числа от 0 до 11: верхняя граница не включается.

```
for i in range(12):
    print(f"{months[i]}: {temps[i]:.1f} °C")
```

Удобнее использовать `zip()`: она объединяет элементы списков попарно,
поэтому обращаться к ним по индексам не нужно. Заодно причешем f-строке `<10` выравнивает текст по левому краю в поле шириной 10, а `>4.1f` — число по правому краю в поле шириной 4, с одним знаком после запятой.

```
for month, temp in zip(months, temps):
    print(f"{month:<10} {temp:>4.1f} °C")
```

Допустим, мы хотим посмотреть только летние месяцы или погоду в последние полгода.
Для этого используем срез: `список[начало:конец]`. Элемент с индексом
`начало` включается, а с индексом `конец` — нет. Июнь имеет индекс 5,
поэтому срез `[5:8]` содержит июнь, июль и август.

```
for month, temp in zip(months[5:8], temps[5:8]):
    print(f"{month:<10} {temp:>4.1f} °C")
```

Если не указать конец среза, он продолжится до конца списка. Например,
`[6:]` выбирает месяцы с июля по декабрь. Аналогично `[:6]` выбирает
элементы с начала списка до индекса 6, не включая его.

```
for month, temp in zip(months[6:], temps[6:]):
    print(f"{month:<10} {temp:>4.1f} °C")
```

### Условия и .методы()
Найдём месяцы, в которых хотя бы в один день средняя температура была ниже
нуля. В нашем словаре для этого достаточно проверить минимум — элемент с
индексом 1. Условие `if` выполняется, когда сравнение истинно, а `else` —
когда оно ложно. С помощью метода`append()` добавим подходящие месяцы в список.

```
cold_months = []
for month in monthly_stats:
    if monthly_stats[month][1] < 0:
        cold_months.append(month)
    else:
        print(f"{month}: дней со средней температурой ниже нуля не было")

cold_months
```

### Создаём отдельную сводную таблицу

Сгруппируем ежедневные средние температуры по месяцу через .groupby() и отдельно вычислим
каждую характеристику. Метод `quantile()` принимает долю от 0 до 1:
например, 0.25 соответствует 25-му процентилю.

```
monthly_temps = weather.groupby("month")["tavg"]
monthly_summary = pd.DataFrame({
    "mean": monthly_temps.mean(),
    "median": monthly_temps.median(),
    "std": monthly_temps.std(),
    "q25": monthly_temps.quantile(0.25),
    "q75": monthly_temps.quantile(0.75),
    "p05": monthly_temps.quantile(0.05),
    "p95": monthly_temps.quantile(0.95),
})
monthly_summary
```

Сохраним таблицу в CSV / Excel / etc

```
monthly_summary.to_csv("monthly-temperature-summary.csv")
```

Прочитаем файл обратно. Заметьте, что теперь столбец month не является индексом

Можно указать `index_col=0`, чтобы первый столбец снова стал индексом,

```
monthly_summary.to_csv("monthly-temperature-summary.csv", index_label="month")
```

## Статистика

Связаны ли средняя температура и количество осадков? Для статистических
тестов используем модуль `stats` из библиотеки SciPy.

```
from scipy import stats
```

Выберем два столбца и удалим строки, в которых хотя бы одно из этих значений
пропущено. Это важно: температура и осадки должны относиться к одному дню.

```
paired = weather[["tavg", "prcp"]].dropna()
paired.shape
```

Выберем два столбца. Их значения расположены в одном порядке: температура
и осадки с одинаковым индексом относятся к одной дате.

```
temperature = paired["tavg"]
precipitation = paired["prcp"]
```

`stats.pearsonr()` вычисляет корреляцию Пирсона — меру линейной связи от −1
до 1 — и p-value для проверки нулевой гипотезы об отсутствии корреляции.
Значение около нуля означает слабую линейную связь, но не обязательно
отсутствие любой связи.

```
correlation = stats.pearsonr(temperature, precipitation)
correlation
```

Как вы понимаете, любой ваш статистический тест из GraphPad или Эксель можно рассчитать и в Python.

### Несколько графиков на одной фигуре

Разместим температуру и осадки друг под другом. `plt.subplots(2, 1)`
создаёт две области графиков в одном столбце, а `sharex=True` объединяет
ось дат. Настроим цвет, толщину линии, прозрачность и сетку.

```
fig, (ax_temp, ax_rain) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

ax_temp.plot(weather.index, weather["tavg"], color="tomato", linewidth=1.5)
ax_temp.axhline(0, color="gray", linestyle="--", linewidth=1)
ax_temp.set_title("Погода в Москве в 2025 году")
ax_temp.set_ylabel("Температура, °C")
ax_temp.grid(alpha=0.3)

ax_rain.bar(weather.index, weather["prcp"], color="steelblue", alpha=0.7)
ax_rain.set_ylabel("Осадки, мм")
ax_rain.set_xlabel("Дата")
ax_rain.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Пунктирная линия отмечает 0 °C, а `tight_layout()` помогает разместить
подписи без наложений. Пропущенные значения осадков не превращаются в нули:
для них столбики не отображаются.
